```python
# Numerical features

numeric_features = [
    "section_distance_km",
    "current_speed_kmph",
    "current_delay_min",
    "historical_avg_section_time_min",
    "congestion_index",
    "speed_restriction",
    "preceding_train_delay_min",
    "rain_intensity",
    "visibility_km",
    "hour",
    "day_of_week",
    "peak_hour"
]


# Categorical features

categorical_features = [
    "train_type",
    "from_station",
    "to_station"
]


# Combine features

all_features = (
    numeric_features +
    categorical_features
)


print("\n================================================")
print("MACHINE LEARNING FEATURES")
print("================================================")

print("\nNumerical Features:")

for feature in numeric_features:
    print(" -", feature)


print("\nCategorical Features:")

for feature in categorical_features:
    print(" -", feature)


print("\nTarget:")
print(" -", target)
```

```python
train_df = pd.read_csv(
    TRAIN_FILE
)

train_df["observation_time"] = pd.to_datetime(
    train_df["observation_time"]
)


validation_df = pd.read_csv(
    VALIDATION_FILE
)

validation_df["observation_time"] = pd.to_datetime(
    validation_df["observation_time"]
)


test_df = pd.read_csv(
    TEST_FILE
)

test_df["observation_time"] = pd.to_datetime(
    test_df["observation_time"]
)


print("\n================================================")
print("TRAIN / VALIDATION / TEST DATA")
print("================================================")

print(
    "Training records :",
    len(train_df)
)

print(
    "Validation records :",
    len(validation_df)
)

print(
    "Testing records :",
    len(test_df)
)
```

```python
X_train = train_df[
    all_features
]

y_train = train_df[
    target
]


X_validation = validation_df[
    all_features
]

y_validation = validation_df[
    target
]


X_test = test_df[
    all_features
]

y_test = test_df[
    target
]


print("\n================================================")
print("DATA SHAPES")
print("================================================")

print("X_train :", X_train.shape)
print("y_train :", y_train.shape)

print("X_validation :", X_validation.shape)
print("y_validation :", y_validation.shape)

print("X_test :", X_test.shape)
print("y_test :", y_test.shape)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            "passthrough",
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)


print("\n================================================")
print("PREPROCESSOR CREATED")
print("================================================")

print(
    "Numerical features -> passed directly"
)

print(
    "Categorical features -> One-Hot Encoded"
)
```

```python
print("\n================================================")
print("HISTORICAL AVERAGE BASELINE")
print("================================================")


section_means = train_df.groupby(
    [
        "from_station",
        "to_station"
    ]
)[target].mean()


global_mean = train_df[
    target
].mean()


baseline_predictions = test_df.apply(
    lambda row:
        section_means.get(
            (
                row["from_station"],
                row["to_station"]
            ),
            global_mean
        ),
    axis=1
)


baseline_mae = mean_absolute_error(
    y_test,
    baseline_predictions
)


baseline_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        baseline_predictions
    )
)


print(
    "Baseline MAE :",
    round(baseline_mae, 2),
    "minutes"
)

print(
    "Baseline RMSE:",
    round(baseline_rmse, 2),
    "minutes"
)
```

```python
print("\n================================================")
print("TRAINING RANDOM FOREST")
print("================================================")


rf_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=300,
                max_depth=18,
                min_samples_leaf=2,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


rf_model.fit(
    X_train,
    y_train
)


print(
    "Random Forest training completed!"
)


rf_predictions = rf_model.predict(
    X_test
)


rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)


rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)


print("\n================================================")
print("RANDOM FOREST RESULTS")
print("================================================")

print(
    "MAE :",
    round(rf_mae, 2),
    "minutes"
)

print(
    "RMSE:",
    round(rf_rmse, 2),
    "minutes"
)
```

```python
print("\n================================================")
print("TRAINING XGBOOST")
print("================================================")


xgb_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            "passthrough",
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)


xgb_model = Pipeline(
    steps=[
        (
            "preprocessor",
            xgb_preprocessor
        ),
        (
            "model",
            XGBRegressor(
                n_estimators=500,
                max_depth=6,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                objective="reg:squarederror",
                random_state=42
            )
        )
    ]
)


xgb_model.fit(
    X_train,
    y_train
)


print(
    "XGBoost training completed!"
)
```

```python
xgb_predictions = xgb_model.predict(
    X_test
)


xgb_mae = mean_absolute_error(
    y_test,
    xgb_predictions
)


xgb_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        xgb_predictions
    )
)


print("\n================================================")
print("XGBOOST RESULTS")
print("================================================")

print(
    "MAE :",
    round(xgb_mae, 2),
    "minutes"
)

print(
    "RMSE:",
    round(xgb_rmse, 2),
    "minutes"
)
```

```python
model_results = pd.DataFrame({
    "Model": [
        "Historical Average",
        "Random Forest",
        "XGBoost"
    ],

    "MAE_minutes": [
        baseline_mae,
        rf_mae,
        xgb_mae
    ],

    "RMSE_minutes": [
        baseline_rmse,
        rf_rmse,
        xgb_rmse
    ]
})


print("\n================================================")
print("MODEL COMPARISON")
print("================================================")

display(
    model_results.round(2)
)


best_model_index = (
    model_results[
        "MAE_minutes"
    ].idxmin()
)


best_model = model_results.loc[
    best_model_index,
    "Model"
]


best_mae = model_results.loc[
    best_model_index,
    "MAE_minutes"
]


print("\n================================================")
print("BEST MODEL")
print("================================================")

print(
    "Best model based on MAE:",
    best_model
)

print(
    "Best MAE:",
    round(best_mae, 2),
    "minutes"
)
```

```python
current_time = pd.Timestamp(
    "2026-09-08 18:40"
)


predicted_eta = (
    current_time
    +
    pd.Timedelta(
        minutes=float(predicted_minutes)
    )
)


print("\n================================================")
print("PREDICTED ETA")
print("================================================")


print(
    "Current time:",
    current_time.strftime(
        "%Y-%m-%d %H:%M"
    )
)


print(
    "Predicted travel time:",
    round(predicted_minutes, 1),
    "minutes"
)


print(
    "Predicted ETA:",
    predicted_eta.strftime(
        "%Y-%m-%d %H:%M"
    )
)
```

```python
normal = live_train.copy()

normal_time = float(
    loaded_model.predict(
        normal
    )[0]
)


normal_eta = (
    current_time
    +
    pd.Timedelta(
        minutes=normal_time
    )
)


disruption = live_train.copy()


disruption["current_speed_kmph"] = 38

disruption["congestion_index"] = 0.92

disruption["speed_restriction"] = 1

disruption["rain_intensity"] = 3

disruption["visibility_km"] = 3


disruption_time = float(
    loaded_model.predict(
        disruption
    )[0]
)


disruption_eta = (
    current_time
    +
    pd.Timedelta(
        minutes=disruption_time
    )
)


print("\n================================================")
print("DISRUPTION SCENARIO")
print("================================================")

print(
    "Predicted travel time:",
    round(disruption_time, 1),
    "minutes"
)

print(
    "Predicted ETA:",
    disruption_eta.strftime("%H:%M")
)


additional_time = (
    disruption_time
    -
    normal_time
)


print("\n================================================")
print("DISRUPTION IMPACT")
print("================================================")

print(
    "Normal travel time:",
    round(normal_time, 1),
    "minutes"
)

print(
    "Disrupted travel time:",
    round(disruption_time, 1),
    "minutes"
)

print(
    "Additional predicted time:",
    round(additional_time, 1),
    "minutes"
)

print(
    "Normal ETA:",
    normal_eta.strftime("%H:%M")
)

print(
    "Disrupted ETA:",
    disruption_eta.strftime("%H:%M")
)
```

```python
def predict_train_eta(
    speed,
    delay,
    congestion,
    rain,
    visibility=5.5,
    speed_restriction=0
):

    scenario = live_train.copy()

    scenario["current_speed_kmph"] = speed

    scenario["current_delay_min"] = delay

    scenario["congestion_index"] = congestion

    scenario["rain_intensity"] = rain

    scenario["visibility_km"] = visibility

    scenario["speed_restriction"] = speed_restriction


    predicted_time = float(
        loaded_model.predict(
            scenario
        )[0]
    )


    eta = (
        current_time
        +
        pd.Timedelta(
            minutes=predicted_time
        )
    )


    return predicted_time, eta
```

```python
# ================================================================
# NORMAL CONDITIONS
# ================================================================

time_normal, eta_normal = predict_train_eta(
    speed=65,
    delay=10,
    congestion=0.30,
    rain=0,
    visibility=10,
    speed_restriction=0
)


print("\n================================================")
print("WHAT-IF SCENARIO 1 — NORMAL")
print("================================================")

print("Speed:", 65, "km/h")
print("Delay:", 10, "minutes")
print("Congestion:", 0.30)
print("Rain:", 0)

print(
    "Predicted travel time:",
    round(time_normal, 1),
    "minutes"
)

print(
    "Predicted ETA:",
    eta_normal.strftime("%H:%M")
)


# ================================================================
# HEAVY DISRUPTION
# ================================================================

time_bad, eta_bad = predict_train_eta(
    speed=38,
    delay=25,
    congestion=0.90,
    rain=3,
    visibility=3,
    speed_restriction=1
)


print("\n================================================")
print("WHAT-IF SCENARIO 2 — HEAVY DISRUPTION")
print("================================================")

print("Speed:", 38, "km/h")
print("Delay:", 25, "minutes")
print("Congestion:", 0.90)
print("Rain:", 3)
print("Speed restriction:", "ON")

print(
    "Predicted travel time:",
    round(time_bad, 1),
    "minutes"
)

print(
    "Predicted ETA:",
    eta_bad.strftime("%H:%M")
)

print(
    "Additional predicted time:",
    round(
        time_bad - time_normal,
        1
    ),
    "minutes"
)
```

```python
# ================================================================
# CONFIDENCE PERCENTAGE + PREDICTION SLOT
# ================================================================

print("\n================================================")
print("PREDICTION CONFIDENCE & TIME SLOT")
print("================================================")

# ------------------------------------------------
# 1. Calculate model error from test predictions
# ------------------------------------------------

absolute_errors = np.abs(
    y_test.values - xgb_predictions
)

mean_error = np.mean(absolute_errors)

# ------------------------------------------------
# 2. Calculate typical prediction uncertainty
# ------------------------------------------------

error_std = np.std(
    y_test.values - xgb_predictions
)

# Approximate prediction interval
lower_bound = max(
    0,
    predicted_minutes - 1.96 * error_std
)

upper_bound = (
    predicted_minutes + 1.96 * error_std
)

# ------------------------------------------------
# 3. Demo confidence score
# ------------------------------------------------
# This is an estimated reliability score,
# NOT predict_proba(), because XGBoost here
# is a regression model.

confidence_score = (
    100 *
    max(
        0,
        1 - (
            error_std /
            max(predicted_minutes, 1)
        )
    )
)

confidence_score = min(
    confidence_score,
    99
)

# ------------------------------------------------
# 4. Prediction slot
# ------------------------------------------------
# Divide ETA into a 10-minute time slot.

slot_start_minute = (
    predicted_eta.minute // 10
) * 10

slot_start = predicted_eta.replace(
    minute=slot_start_minute,
    second=0,
    microsecond=0
)

slot_end = (
    slot_start +
    pd.Timedelta(minutes=10)
)

# ------------------------------------------------
# 5. Display results
# ------------------------------------------------

print(
    "Predicted travel time:",
    round(predicted_minutes, 2),
    "minutes"
)

print(
    "Estimated confidence:",
    round(confidence_score, 1),
    "%"
)

print(
    "Prediction interval:",
    round(lower_bound, 1),
    "to",
    round(upper_bound, 1),
    "minutes"
)

print(
    "Predicted ETA:",
    predicted_eta.strftime(
        "%Y-%m-%d %H:%M"
    )
)

print(
    "Prediction slot:",
    slot_start.strftime("%H:%M"),
    "-",
    slot_end.strftime("%H:%M")
)

print(
    "================================================"
)
```
